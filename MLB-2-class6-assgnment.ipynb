# Merging and joining

```python
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 116, 117, 118, 119, 120],
    "name": ["Rahim", "Karim", "Hasan", "Sakib", "Nabil",
             "Tanvir", "Fahim", "Rafi", "Imran", "Shakil",
             "Arif", "Jahid", "Mehedi", "Siam", "Tareq",
             "Rakib", "Adnan", "Nayeem", "Foysal", "Rony"],
    "department": ["CSE", "EEE", "CSE", "BBA", "CSE",
                   "EEE", "CSE", "BBA", "CSE", "EEE",
                   "CSE", "BBA", "EEE", "CSE", "BBA",
                   "CSE", "EEE", "CSE", "BBA", "EEE"]
})

students
```

```python
results = pd.DataFrame({
    "student_id": [101, 102, 103, 105, 106, 108, 109, 111, 112, 114,
                   121, 122, 123, 124, 125, 126, 127, 128, 129, 130],
    "math": [85, 72, 91, 68, 77, 88, 74, 95, 81, 69,
             78, 84, 92, 73, 89, 67, 80, 94, 76, 83],
    "python": [90, 75, 88, 72, 80, 92, 79, 96, 85, 71,
               82, 87, 95, 76, 91, 70, 84, 93, 78, 86]
})

results
```

```python
departments = pd.DataFrame({
    "dept_id": [1, 2, 3, 4],
    "department": [
        "CSE",
        "EEE",
        "BBA",
        "Civil"
    ],
    "faculty": [
        "Engineering",
        "Engineering",
        "Business",
        "Engineering"
    ]
})

departments
```

# 1. Create a dataframe contains students name,faculty and number on math and python

```python
data1=students.merge(departments, how='inner' )
data1
```

```python
data2=data1.merge(results)
data2
```

```python
data3=data2.loc[:,['name','faculty','math','python']]
data3
```

# 6. Calculate month-wise average windspeed

```python
climate['month']=climate.index.month_name()
climate['year']=climate.index.year
avg=climate.groupby(['year','month'])['wind_speed'].mean()
avg
```

```python
monthly=climate['wind_speed'].resample('ME').mean()
monthly
```

# 7. Calculate month-wise highest meantemp

```python
mt=climate.groupby(['year','month'])['meantemp'].max()
mt.plot(kind='bar')
```

# 8. Year-wise maximum, minimum and average humidity

```python
climate.groupby('year')['humidity'].agg(['max','min','mean'])
```

# 9. A new column contains previous 10 days average meantemp

```python
climate['temp_prev_10']=climate['meantemp'].rolling(10).mean()
climate.tail(10)
```
